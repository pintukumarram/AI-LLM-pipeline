# 01. Introduction to Python

## 1.1 What is Python?

Python is a high-level, general-purpose, interpreted programming language known for its simple syntax, readability, and extensive ecosystem of libraries.

Python supports multiple programming paradigms, including:
- Procedural Programming
- Object-Oriented Programming (OOP)
- Functional Programming

### Key Features of Python

1. **Simple and Readable:** Python has a clean and understandable syntax.
2. **High-Level Language:** Developers do not need to manage low-level hardware details directly.
3. **Dynamically Typed:** Variable types are determined at runtime.
4. **Interpreted Execution:** Python code is executed through an interpreter environment.
5. **Object-Oriented:** Python supports classes, objects, inheritance and polymorphism.
6. **Cross-Platform:** Python programs can run on different operating systems.
7. **Extensive Libraries:** Python provides a large standard library and third-party ecosystem.
8. **Automatic Memory Management:** Python manages memory automatically.

### Applications of Python

- Web Development
- Data Science and Analytics
- Artificial Intelligence and Machine Learning
- Automation and Scripting
- Backend Development
- Cybersecurity
- Scientific Computing



## 1.2 How does Python execute a program?

Python execution involves several stages.

1. Source code is written in a `.py` file.
2. Python parses and compiles the source code into code objects.
3. CPython commonly generates bytecode.
4. The Python Virtual Machine (PVM) executes the bytecode.
5. Results are produced at runtime.

**Important:** Python is commonly called an interpreted language, but CPython also performs a compilation-to-bytecode step before execution.

A variable is a name that refers to an object in Python.

Unlike some statically typed languages, Python does not require an explicit type declaration when creating a variable.

### Syntax

```python

```

### Example

```python

```

### Important points

2. Python is dynamically typed.
3. A variable can be reassigned to an object of another type.

## 2.4 Type Conversion in Python

Type conversion means converting a value from one data type to another.

Python supports two major types of conversion.

### 1. Implicit Type Conversion

Python automatically converts a value into another compatible type during certain operations.

Example:

```python
a = 10
b = 2.5

result = a + b

print(result)
print(type(result))
```

Output:

```text
12.5
<class 'float'>
```

Here, Python promotes the integer to a float for the arithmetic operation.

### 2. Explicit Type Conversion

Explicit conversion is performed by the programmer using built-in functions.

Common conversion functions:

| Function | Purpose |
|---|---|
| int() | Convert to integer |
| float() | Convert to float |
| str() | Convert to string |
| bool() | Convert to boolean |
| list() | Convert to list |
| tuple() | Convert to tuple |
| set() | Convert to set |
| dict() | Create a dictionary from suitable key-value pairs |

**Important:** Not every conversion is valid. For example, `int("hello")` raises a ValueError.


## Checking data types

### 2.5 type() and isinstance()

Python provides two useful functions for checking object types.

### type()

Returns the exact type of an object.

### isinstance()

Checks whether an object is an instance of a specified class or its subclass.

Example:

```python
number = 100

print(type(number))
print(isinstance(number, int))
```

Output:

```text
<class 'int'>
True
```

### Difference

- `type()` checks the exact type.
- `isinstance()` supports inheritance-aware type checking.

In production code, `isinstance()` is often more flexible when working with class hierarchies.


````python
# Integer to float
number = 10
converted_float = float(number)

# Float to integer
price = 99.99
converted_int = int(price)

# Integer to string
age = 23
converted_string = str(age)

# String to integer
value = "500"
converted_value = int(value)

print(converted_float, type(converted_float))
print(converted_int, type(converted_int))
print(converted_string, type(converted_string))
print(converted_value, type(converted_value))
````

````python
#type() vs isinstance()
class Animal:
    pass


class Dog(Animal):
    pass


dog = Dog()

print(type(dog) == Dog)
print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
````

## Mutable and immutable objects


### 2.6 Mutable vs Immutable Data Types

This is one of the most important Python interview concepts.

### Mutable Objects

Mutable objects can be modified after creation without creating a new object.

Examples:

- list
- dict
- set
- bytearray

### Immutable Objects

Immutable objects cannot be modified after creation.

Examples:

- int
- float
- bool
- str
- tuple
- frozenset
- bytes

### Example

```python
numbers = [10, 20, 30]
numbers.append(40)

print(numbers)
```

The existing list is modified.

Now consider a string:

```python
name = "Python"
name = name + " Programming"

print(name)
```

A new string object is created for the concatenated result, and the variable is rebound to it.

**Interview point:** Mutability is a property of objects, not variable names.

## 3.1 Types of Operators

Python provides the following operators:

| Type | Operators |

### Important tricky operators

- `/` performs division and returns a float.
- `//` performs floor division.
- `%` returns the remainder.
- `**` performs exponentiation.
- `==` compares values.
- `is` compares object identity.

### Tricky operators


````python
print(10 / 3)     # 3.333...
print(10 // 3)    # 3
print(-10 // 3)   # -4
print(10 % 3)     # 1
print(2 ** 3)     # 8

print(5 == 5)     # True
print(5 is 5)     # True (implementation may reuse this object)
````

Important: // rounds towards negative infinity, not towards zero.

Conditional statements execute code based on conditions.

### Types

## Loops in Python

Loops execute a block of code repeatedly.

### for loop
Used to iterate over an iterable.

### while loop
Executes while a condition remains True.

### Loop control statements

- break: Terminates the loop.
- continue: Skips the current iteration.
- pass: Does nothing; acts as a placeholder.

### Loop control

````python
for i in range(1, 6):

    if i == 3:
        continue

    if i == 5:
        break

    print(i)
````

## Important Tricky concept

1. Truthy and Falsy Values

Falsy values include:
- False
- None
- 0
- 0.0
- ""
- []
- {}
- set()

Most other values are truthy.

### 2. Short-Circuit Evaluation

Python evaluates logical expressions from left to right.

- In `and`, evaluation stops when a falsy value is found.
- In `or`, evaluation stops when a truthy value is found.

### 3. Operator Precedence

Operators follow precedence rules.

Example:
2 + 3 * 4 = 14

Multiplication happens before addition.

Use parentheses to make expressions explicit.

### Short-circuit evaluation

````python
x = 0

result = x and (10 / x)

print(result)

result = 10 or (10 / 0)

print(result)
````
